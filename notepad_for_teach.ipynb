```python
from ultralytics import YOLO

# 1. Ładujemy bazowy, najszybszy model Nano
model = YOLO('yolov8n.pt')

# 2. Startujemy trening
model.train(
    data='/content/drive/MyDrive/model_ropoznawainie_ręki/data.yaml',

    epochs=70,           # 70 epok
    imgsz=640,
    patience=20,         # Jeśli przez 20 epok nie będzie rekordu, kończymy

    degrees=40.0,        # Ręka pod różnymi kątami
    translate=0.2,       # Ręka w różnych miejscach kadru
    scale=0.6,           # Ręka blisko i daleko
    fliplr=0.5,          # Lustrzane odbicie
    mosaic=1.0,          # Szukanie dłoni w bałaganie
    hsv_v=0.4,           # Odporność na ciemność/jasność
    hsv_s=0.7,           # Odporność na różne kolory skóry

    optimizer='AdamW',
    cos_lr=True,         # Łagodne wygaszanie nauki (wyższa precyzja na koniec)
    verbose=True         # Chcemy widzieć postępy
)
```
